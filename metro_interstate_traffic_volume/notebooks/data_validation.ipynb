# Kiểm tra chất lượng dữ liệu giao thông  
Mục tiêu:  
- Kiểm tra kích thước và kiểu dữ liệu.
- Kiểm tra giá trị thiếu.
- Kiểm tra dòng trùng và timestamp trùng.
- Xác định các giờ bị thiếu.
- Phát hiện giá trị bất hợp lý.  
Dữ liệu: Metro Interstate Traffic Volume.

## Import thư viện

```python
from pathlib import Path
import pandas as pd
```

## xác định đường dẫn 

```python

cwd = Path.cwd()

if (cwd / "data").is_dir():
    PROJECT_DIR = cwd
elif (cwd.parent / "data").is_dir():
    PROJECT_DIR = cwd.parent
else:
    raise FileNotFoundError("Không tìm thấy thư mục data của dự án.")

DATA_PATH = PROJECT_DIR / "data" / "Metro_Interstate_Traffic_Volume.csv"

print("Đường dẫn dữ liệu:", DATA_PATH)
print("File tồn tại:", DATA_PATH.is_file())
```

## Đọc và xem dữ liêu

```python
df_raw = pd.read_csv(
    DATA_PATH,
    keep_default_na=False
)

print(f"Số dòng: {df_raw.shape[0]:,}")
print(f"Số cột: {df_raw.shape[1]}")

display(df_raw.head())
```
