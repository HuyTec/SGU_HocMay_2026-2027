## PHẦN 1: KHÁM PHÁ VÀ PHÂN PHỐI SALEPRICE

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### 1. Đọc dữ liệu và Kiểm tra kích thước

```python
train = pd.read_csv("data/train.csv")
test = pd.read_csv("data/test.csv")
print("Train:", train.shape)
print("Test :", test.shape)
```

### 2. Xem dữ liệu

```python
print("\n5 dòng đầu tiên:")
print(train.head())
```

### 3. Thống kê SalePrice

```python
X = train.drop("SalePrice", axis=1)
y = train["SalePrice"]
train["SalePrice"].describe()
```

#### 4. Skewness trước log

```python
train["SalePrice"].skew()
```

### 5. Vẽ phân phối SalePrice

```python
plt.figure(figsize=(10, 6))

sns.histplot(train["SalePrice"], kde=True)

plt.title("Distribution of SalePrice")
plt.xlabel("SalePrice")
plt.ylabel("Frequency")

plt.show()
```

### 6. Log transformation

```python
train["LogSalePrice"] = np.log1p(train["SalePrice"])
train["LogSalePrice"].head()
```

### 7. Skewness sau log

```python
train["LogSalePrice"].skew()
```

### 8. Vẽ phân phối sau log

```python
plt.figure(figsize=(10, 6))

sns.histplot(train["LogSalePrice"], kde=True)

plt.title("Log(SalePrice) Distribution")
plt.xlabel("Log(SalePrice)")
plt.ylabel("Frequency")

plt.show()
```

### 9. So sánh

```python
print("Skewness trước log:",
      train["SalePrice"].skew())

print("Skewness sau log:",
      train["LogSalePrice"].skew())
```
