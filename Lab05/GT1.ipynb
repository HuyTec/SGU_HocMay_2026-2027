# 1 GIẢI THUẬT 1: CÂY QUYẾT ĐỊNH VÀ RỪNG CÂY
## 1.1 ôn tập lý thuyết


- Quy trình khai phá dữ liệu CRISP – DM (Cross Industry Standard Process for Data Mining) là gì ? Quy
trình khai phá dữ liệu SEMMA (Sample, Explore, Modify, Model, Access) là gì?  
CRISP-DM  là quy trình chuẩn liên ngành cho khai phá dữ liệu.  
SEMMA là quy trình khai phá dữ liệu của SAS
- Cây quyết định hoạt động như thế nào? Hãy giải thích các thành phần chính (nút gốc, nút lá, nhánh)
và cách cây đưa ra dự đoán.  
Mô hình đưa ra dự đoán bằng cách lần lược kiểm tra các điều kiện của dữ liệu, nó hoạt động giống một chuỗi câu hỏi mỗi hỏi rẽ sang 1 nhánh đến khi  
Nút gốc: nút đầu tiên của cây chứa điều kiện kiểm tra ban đầu  
Nút Trong: chứa các điều kiện tiếp theo để tiếp tục chia dữ liệu thành những nhóm nhỏ hơn.  
nhánh: đường nối giữa các nút biểu hiện kết quả của một điều kiện
lá: nút kết thúc , chứa kết quản dự đoán: một nhãn phân loại hoặc một giá trị số.
- Các tiêu chí phân tách (splitting criteria) như Gini Index, Entropy, hay Information Gain được sử dụng
trong cây quyết định là gì? Chúng khác nhau ra sao?
Gini là xác xuất hai mẫu được độc lập theo phân bố lớp có mẫu khác nhau
Entropy : đo mức độ khó xác định lớp của một mẫu tại nút.  
Information Gain: đo entropy giảm được bao nhiêu sau phép chia.
