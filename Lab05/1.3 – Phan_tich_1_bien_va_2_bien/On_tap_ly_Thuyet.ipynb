### Câu 1

**Câu hỏi:** Phân tích đơn biến và hai biến trong EDA là gì? Chúng khác nhau như thế nào?

**Trả lời:**

**Phân tích đơn biến (univariate analysis)** xem xét từng biến riêng lẻ để mô tả phân phối, giá trị phổ biến, độ phân tán, giá trị thiếu hoặc ngoại lệ. Ví dụ: phân tích phân phối `Income` của khách hàng bằng histogram.

**Phân tích hai biến (bivariate analysis)** xem xét đồng thời **hai biến** để tìm mối liên hệ hoặc sự khác nhau theo nhóm. Ví dụ: thu nhập `Income` và số tiền mua rượu `MntWines` bằng scatterplot. Phân tích hai biến không đồng nghĩa với kết luận nguyên nhân.

| Tiêu chí | Đơn biến | Hai biến |
|---|---|---|
| Số biến | Một | Hai |
| Mục tiêu | Hiểu phân phối một thuộc tính | Hiểu quan hệ/khác biệt giữa hai thuộc tính |
| Ví dụ | Histogram của `Income` | Scatterplot `Income`–`MntWines` |

### Câu 2

**Câu hỏi:** Các thước đo thống kê nào thường sử dụng trong phân tích đơn biến?

**Trả lời:**

- **Mean (trung bình):** tổng các giá trị chia số quan sát hợp lệ; dễ chịu tác động của ngoại lệ.
- **Median (trung vị):** giá trị ở vị trí giữa sau khi sắp xếp; bền vững hơn trước ngoại lệ.
- **Mode (yếu vị):** giá trị xuất hiện phổ biến nhất, phù hợp với biến định tính và định lượng rời rạc.
- **Variance (phương sai):** trung bình bình phương độ lệch quanh mean, đo phân tán theo đơn vị bình phương.
- **Standard deviation (độ lệch chuẩn):** căn bậc hai phương sai, có cùng đơn vị với dữ liệu.
- **Q1, Q3, IQR:** các phân vị 25%, 75% và khoảng tứ phân vị `IQR = Q3 - Q1`; hỗ trợ nhận diện ngoại lệ.

Với `Income` có giá trị rất lớn so với phần lớn quan sát, nên xem **median và IQR** bên cạnh mean, không tùy tiện xóa giá trị cao.

### Câu 3

**Câu hỏi:** Trong phân tích hai biến, làm sao xác định quan hệ giữa hai biến? Tương quan khác nhân quả thế nào?

**Trả lời:**

- Với **hai biến số**, vẽ scatterplot và kiểm tra tương quan Pearson (quan hệ **tuyến tính**) hoặc Spearman (quan hệ đơn điệu theo thứ hạng).

**Tương quan không chứng minh nhân quả.** Hai biến có thể đồng biến vì biến thứ ba, xu thế thời gian hoặc sai lệch chọn mẫu. Để bàn luận quan hệ nhân quả, cần thiết kế nghiên cứu và giả định thích hợp; scatterplot chỉ cho biết hình dạng liên hệ trong dữ liệu.

### Câu 4

**Câu hỏi:** Tương quan (correlation) và hiệp biến (covariance) khác nhau ra sao?

**Trả lời:**

**Covariance (hiệp biến)** cho biết dấu biến thiên cùng chiều (`>0`) hay ngược chiều (`<0`) giữa hai biến. Giá trị của nó thay đổi theo đơn vị đo, nên không dễ so mức mạnh yếu giữa các cặp.

**Pearson correlation (hệ số tương quan Pearson)** chuẩn hóa covariance theo độ lệch chuẩn:

$$r_{XY}=\frac{\operatorname{Cov}(X,Y)}{s_X s_Y}$$

Khi xác định, `r` nằm trong `[-1,1]`: gần 1 là tuyến tính dương mạnh; gần -1 là tuyến tính âm mạnh; gần 0 là ít quan hệ tuyến tính (nhưng vẫn có thể có quan hệ phi tuyến). Tương quan không có đơn vị.

### Câu 5

**Câu hỏi:** Khi nào sử dụng biểu đồ đơn biến so với hai biến?

**Trả lời:**

| Kiểu biến | Mục đích | Biểu đồ phù hợp |
|---|---|---|
| Một biến số | Phân phối, lệch, ngoại lệ | Histogram, boxplot, violinplot |
| Một biến phân loại | Đếm và tỷ lệ từng nhóm | Countplot, bar chart, pie chart (ít nhóm) |
| Hai biến số | Xu hướng liên hệ | Scatterplot, đường hồi quy, heatmap tương quan |
| Hai biến phân loại | Tỷ lệ theo nhóm | Crosstab, heatmap tần suất |
| Một biến số + một biến phân loại | So sánh phân phối giữa các nhóm | Boxplot/violinplot theo nhóm |

Ví dụ PDF: histogram chiều dài mỏ (đơn biến), scatterplot chiều dài mỏ–khối lượng chim cánh cụt (hai biến).

### Câu 6

**Câu hỏi:** Ví dụ mã Python cho scatterplot và heatmap dùng để phân tích hai biến?

**Trả lời:**

Ví dụ dùng tập dữ liệu **Iris** có sẵn trong `scikit-learn` để người học chạy được ngay, không phụ thuộc file CSV. Scatterplot cho thấy quan hệ giữa chiều dài và chiều rộng cánh hoa; heatmap minh họa ma trận hệ số tương quan giữa các biến số.

Mã Python ở ô tiếp theo. Đọc biểu đồ phải nói rõ đây là **mối liên hệ quan sát được**, không phải bằng chứng nhân quả.

```python
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

iris = load_iris(as_frame=True)
df_iris = iris.frame.rename(columns={"target": "species"}).copy()
df_iris["species"] = df_iris["species"].map(dict(enumerate(iris.target_names)))
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df_iris, x="petal length (cm)", y="petal width (cm)", hue="species")
plt.title("Iris: hai biến số (scatterplot)")
plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 4))
sns.heatmap(df_iris.select_dtypes("number").corr(), annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Ma trận tương quan Pearson (Iris)")
plt.tight_layout()
plt.show()
```

### Câu 7

**Câu hỏi:** Làm sao trực quan hóa quan hệ giữa biến số và biến phân loại bằng boxplot hoặc violin plot?

**Trả lời:**

Ví dụ với Iris: `species` là **biến phân loại**, `petal length (cm)` là **biến số**.

- `sns.boxplot()` hiển thị trung vị, khoảng tứ phân vị và các điểm nằm ngoài râu hộp theo quy tắc biểu đồ.
- `sns.violinplot()` hiển thị hình dạng phân bố (mật độ ước lượng) cho từng nhóm, giúp thấy vùng tập trung và độ trải.

Đặt `x='species'`, `y='petal length (cm)'` để so sánh ba loài hoa. Không diễn giải sự khác biệt giữa các nhóm như tác động nhân quả.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.boxplot(data=df_iris, x="species", y="petal length (cm)", ax=axes[0])
axes[0].set_title("Boxplot theo loài")
sns.violinplot(data=df_iris, x="species", y="petal length (cm)", ax=axes[1])
axes[1].set_title("Violinplot theo loài")
plt.tight_layout()
plt.show()
```
